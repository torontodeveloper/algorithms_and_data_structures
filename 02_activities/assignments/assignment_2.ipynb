```python
import hashlib

def hash_to_range(input_string: str) -> int:
     hash_object = hashlib.sha256(input_string.encode())
     hash_int = int(hash_object.hexdigest(), 16)
     return (hash_int % 3) + 1
# Kevin
input_string = input("Plz enter your first name")
result = hash_to_range(input_string)
print(result)
```

```python

def is_duplicate(root: TreeNode):
    print(root.val,end=" ")
    global isFound
    if root is None:
        return -1
    if result.get(root.val):
        print("Duplicate found",root.val)
        isFound = True
        return root.val
    result[root.val]=1
    if (not isFound) and root.left is not None:
        is_duplicate(root.left)

    if (not isFound)  and root.right is not None:
        is_duplicate(root.right)
    if not isFound:
        return -1
```

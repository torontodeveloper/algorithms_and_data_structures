#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val = 0,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right
        
```

```python
from collections import deque
def is_duplicate(root: TreeNode):
    if root is None:
        return -1
    queue = deque([root])
    seen_values = set()
    while queue:
        node = queue.popleft()
        if node.val in seen_values:
            return node.val
        seen_values.add(node.val)
        if node.left:
            queue.append(node.left)
        if node.right:
              queue.append(node.right)
    return -1
```

```python
#  [1, 2, 2, 3, 5, 6, 7]
root = TreeNode(1)
node21 = TreeNode(2)
node22 = TreeNode(2)
node3 = TreeNode(3)
node5 = TreeNode(5)
node6 = TreeNode(6)
node7 = TreeNode(7)
root.left = node21
root.right = node22
root.left.left = node3
root.left.right = node5
root.right.left = node6
root.right.right = node7
result = {}
isFound= False
is_duplicate(root)
```
